```python
import requests
from bs4 import BeautifulSoup
```

```python
page_count = 1 # Start from page 1

while True :
      url = f"http://quotes.toscrape.com/page/{page_count}/"
      res = requests.get(url)

      soup = BeautifulSoup(res.text , "lxml")
      quotes = soup.select("div.quote")


      if  not quotes : # If no quotes are found, it means we've reached the end
        print(f"No valid page found at {url}. Stopping scrape.")
        break

      with open(f"sample_data/quotes{page_count}.html","w") as f :
        f.write(res.text)
        print(f"Downloaded data from page {page_count}")

      page_count +=1 # Increment page count for the next iteration
```

```python
#  extract  the imp info
with open("sample_data/quotes1.html", "r") as f :
  html_content = f.read()

soup = BeautifulSoup(html_content , "lxml")

all_quotes = soup.select("div.quote")
```

```python
for quote in all_quotes :
  all_tag = quote.select(".tags .tag")[0].get_text(strip=True)
  print(all_tag)
```

```python
for quote in all_quotes :
  all_tag =  []
  for tag in quote.select(".tags .tag"):
    all_tag.append(tag.get_text())
  print(all_tag)
```

```python
life_quotes = []

for quote in all_quotes:
  # Extract all tags for the current quote
  current_quote_tags = [tag.get_text(strip=True) for tag in quote.select(".tags .tag")]

  if "life" in current_quote_tags:
    text = quote.select_one("span.text").get_text()
    author = quote.select_one("small.author").get_text()
    life_quotes.append([text, author])

print("Quotes with 'life' tag:")
for q_text, q_author in life_quotes:
    print(f"- {q_text} by {q_author}")
```

### Extracting Quotes from All Downloaded Pages (1-10)

Now, let's loop through all the saved HTML files (from `quotes1.html` to `quotes10.html`), parse their content, and extract the quote text, author, and associated tags from each quote. We will store all this information in a list of dictionaries, which can then be converted into a pandas DataFrame for easier analysis.

```python
all_scraped_quotes_data = []

for page_num in range(1, 11): # Loop from page 1 to 10
    file_path = f"sample_data/quotes{page_num}.html"

    try:
        with open(file_path, "r", encoding="utf-8") as f:
            html_content = f.read()

        soup = BeautifulSoup(html_content, "lxml")
        quotes_on_page = soup.select("div.quote")

        if not quotes_on_page:
            print(f"No quotes found on {file_path}. Skipping.")
            continue

        for quote in quotes_on_page:
            text = quote.select_one("span.text").get_text(strip=True)
            author = quote.select_one("small.author").get_text(strip=True)
            tags = [tag.get_text(strip=True) for tag in quote.select(".tags .tag")]

            all_scraped_quotes_data.append({
                'page': page_num,
                'text': text,
                'author': author,
                'tags': tags
            })
        print(f"Processed quotes from {file_path}")

    except FileNotFoundError:
        print(f"File not found: {file_path}. It might not have been downloaded.")
    except Exception as e:
        print(f"An error occurred while processing {file_path}: {e}")

print(f"Total quotes collected: {len(all_scraped_quotes_data)}")
```

```python
import pandas as pd

# Convert the list of dictionaries to a pandas DataFrame
df_quotes = pd.DataFrame(all_scraped_quotes_data)

# Display the first 5 rows of the DataFrame
print("DataFrame of all scraped quotes:")
display(df_quotes.head())

# Display some basic info about the DataFrame
print("\nDataFrame Info:")
df_quotes.info()
```

```python
import requests
from bs4 import BeautifulSoup

url = "http://books.toscrape.com/catalogue/category/books/travel_2/index.html"

try:
    # Send a GET request to the URL
    response = requests.get(url)

    # Check if the request was successful
    if response.status_code == 200:
        soup = BeautifulSoup(response.text, 'html.parser')

        # Find all book items (usually inside <article> tags with class 'product_pod')
        books = soup.find_all('article', class_='product_pod')

        if not books:
            print("No books found on this page.")
        else:
            print(f"Found {len(books)} books:\n")
            print("-" * 50)

            for book in books:
                # Extract Name (Title)
                # The title is in the 'title' attribute of the <a> tag inside <h3>
                title_tag = book.find('h3').find('a')
                name = title_tag.get('title')

                # Extract Price
                # The price is in the 'p' tag with class 'price_color'
                price_tag = book.find('p', class_='price_color')
                price = price_tag.text if price_tag else "N/A"

                # Extract Author
                # Note: On books.toscrape.com category pages, the author name is NOT present in the HTML.
                # It is only available on the individual book detail page.
                # We will print a note indicating this limitation.
                author = "Author not listed on category page (requires detail page)"

                # Optional: Extract Star Rating (often present)
                star_rating = book.find('p', class_='star-rating')
                if star_rating:
                    rating_class = star_rating.get('class')
                    rating = rating_class if len(rating_class) > 1 else "N/A"
                else:
                    rating = "N/A"

                print(f"Name: {name}")
                print(f"Price: {price}")
                print(f"Rating: {rating}")
                print(f"Author: {author}")
                print("-" * 50)

    else:
        print(f"Failed to retrieve the page. Status code: {response.status_code}")

except requests.exceptions.RequestException as e:
    print(f"An error occurred while fetching the page: {e}")
```
