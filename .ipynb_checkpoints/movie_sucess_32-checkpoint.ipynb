```python
import pandas as pd 
from sklearn.tree import DecisionTreeClassifier
# from sklearn.preprocessing import train_test_split 
from sklearn.model_selection import train_test_split
# from sklearn.matrix import accuracy_score , calssification_report 
```

```python
df = pd.read_csv(r"ml/supervied ml/movie_sucess.csv")
```

```python
df.head()
```

```python
df.info()
df.isnull().sum()          # count of nulls per column
df.columns[df.isnull().any()]   # names of columns with any null
df = df.dropna(axis=0)
df = df.dropna(axis=1)
```

```python
import seaborn as sns
sns.heatmap(df.isnull())
```

```python
sns.boxplot(y=df['Revenue (Millions)'])
```

```python
import matplotlib.pyplot as plt

num_cols = ['Rating', 'Revenue (Millions)', 'Year']

plt.figure(figsize=(12,4))
for i, col in enumerate(num_cols):
    plt.subplot(1, len(num_cols), i+1)
    sns.boxplot(y=df[col])
    plt.title(col)
plt.tight_layout()
plt.show()
```

```python
import numpy as np

df['Revenue (Millions)'] = np.log1p(df['Revenue (Millions)'])
```

```python
sns.boxplot(y=df['Revenue (Millions)'])
plt.title('Revenue (Log Transformed)')
plt.show()
```

```python
X = df.drop(['Success', 'Title', 'Genre', 'Description', 'Director', 'Actors'] , axis = 1)
y = df['Success']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
```

```python
model = DecisionTreeClassifier()
model.fit(X_train,y_train)
```

```python

y_pred =  model.predict(X_test)
```

```python
y_pred
```

```python
X_test
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
print("Train accuracy:", model.score(X_train, y_train))
print("Test accuracy:", model.score(X_test, y_test))
```

```python
X = df.drop(columns=['Success', 'Title', 'Genre', 'Description', 'Director', 'Actors', 
                      'Revenue (Millions)', 'Revenue_log', 'Votes'])
y = df['Success']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

model = DecisionTreeClassifier(max_depth=5, random_state=42)  # also cap depth to avoid overfitting
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
print("Train accuracy:", model.score(X_train, y_train))
print("Test accuracy:", model.score(X_test, y_test))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
model = DecisionTreeClassifier(max_depth=5, class_weight='balanced', random_state=42)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
print("Train accuracy:", model.score(X_train, y_train))
print("Test accuracy:", model.score(X_test, y_test))
print(classification_report(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```
