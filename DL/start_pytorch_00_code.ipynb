#reproducbillity (trying to take random out of random)

* in short how neural network works learns:
start with randoms-->tensoroperation-->update random no. to try and make them better representation of the data -->again-->again-->again...

```python
### to reduce the randomnes in pytorch comes the concept of the **random seed** essentially WHAT THE RANDOM  seed does is "flavour" the randomness
```

```python
torch.rand(3,3)
```

```python
# creating 2 random tensor
random_tensor_A = torch.rand(3,4)
random_tensor_B = torch.rand(3,4)
print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```

```python
# lets make some random reproduceble tensor
# set the random seed
random_seed = 42
torch.manual_seed(random_seed)
random_tensor_C = torch.rand(3,4)

torch.manual_seed(random_seed)
random_tensor_D =torch.rand(3,4)
print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C == random_tensor_D)
#
```

```python
torch.manual_seed(random_seed)
```
